```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.datasets import make_classification
from mpl_toolkits.mplot3d import Axes3D
import os
import cv2
from sklearn.model_selection import train_test_split
#from sklearn.preprocessing import LabelEncode
from sklearn.metrics import accuracy_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
import time
```

```python
import os
import cv2
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

dataset_path = '/content/my_dataset/Brain_Tumor'
categories = ['Grayscale_Brain_Healthy', 'Grayscale_Brain_Tumor']

data = []
labels = []

for category in categories:
    folder_path = os.path.join(dataset_path, category)
    for img_name in os.listdir(folder_path):
        img_path = os.path.join(folder_path, img_name)
        try:
            img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
            data.append(img)
            labels.append(category)
        except Exception as e:
            print(f"Error loading image {img_path}: {e}")

data = np.array(data)
labels = np.array(labels)

print("Data shape:", data.shape)
print("Labels shape:", labels.shape)

le = LabelEncoder()
labels_encoded = le.fit_transform(labels)

X_train, X_test, y_train, y_test = train_test_split(
    data, labels_encoded, test_size=0.3, random_state=42, stratify=labels_encoded
)

X_train = X_train / 255.0
X_test = X_test / 255.0

X_train_flat = X_train.reshape(X_train.shape[0], -1)
X_test_flat = X_test.reshape(X_test.shape[0], -1)

print("Flattened train shape:", X_train_flat.shape)
print("Flattened test shape:", X_test_flat.shape)
pca = PCA(n_components=50)
X_train_pca = pca.fit_transform(X_train_flat)
X_test_pca = pca.transform(X_test_flat)

print("PCA train shape:", X_train_pca.shape)
```

```python


def evaluate_classifiers_with_pca_and_plot(X_train_flat, y_train, X_test_flat, y_test):
    # PCA with 100 components
    pca = PCA(n_components=100)
    X_train_pca = pca.fit_transform(X_train_flat)
    X_test_pca = pca.transform(X_test_flat)

    knn_accuracies = []
    knn_times = []
    k_values = [1, 3, 5, 7]

    print("KNN Classifier Performance:")
    for k in k_values:
        knn = KNeighborsClassifier(n_neighbors=k)
        start = time.time()
        knn.fit(X_train_pca, y_train)
        end = time.time()

        y_pred = knn.predict(X_test_pca)
        acc = accuracy_score(y_test, y_pred)
        knn_accuracies.append(acc)
        knn_times.append(end - start)

        print(f"  k={k}: Accuracy = {acc:.4f}, Training Time = {end - start:.4f} seconds")

    print("\nDecision Tree Classifier:")
    dt = DecisionTreeClassifier(random_state=42)
    start = time.time()
    dt.fit(X_train_pca, y_train)
    end = time.time()
    y_pred_dt = dt.predict(X_test_pca)
    acc_dt = accuracy_score(y_test, y_pred_dt)
    print(f"  Accuracy = {acc_dt:.4f}, Training Time = {end - start:.4f} seconds")

    print("\nSVM Classifier:")
    svm = SVC(random_state=42)
    start = time.time()
    svm.fit(X_train_pca, y_train)
    end = time.time()
    y_pred_svm = svm.predict(X_test_pca)
    acc_svm = accuracy_score(y_test, y_pred_svm)
    print(f"  Accuracy = {acc_svm:.4f}, Training Time = {end - start:.4f} seconds")
    print(f"  Number of Support Vectors = {sum(svm.n_support_)}")


    plt.figure(figsize=(10, 6))
    plt.plot(k_values, knn_accuracies, marker='o', label='KNN Accuracy')
    plt.axhline(y=acc_dt, color='r', linestyle='--', label='Decision Tree Accuracy')
    plt.axhline(y=acc_svm, color='g', linestyle='--', label='SVM Accuracy')

    plt.title('Classifier Accuracy at 100 PCA Components')
    plt.xlabel('K value for KNN')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.grid(True)
    plt.show()

evaluate_classifiers_with_pca_and_plot(X_train_flat, y_train, X_test_flat, y_test)

```

```python
from sklearn.cluster import DBSCAN

X_train_class0 = X_train_pca[y_train == 0]  # Class 0: e.g., Grayscale_Brain_Healthy
X_train_class1 = X_train_pca[y_train == 1]  # Class 1: e.g., Grayscale_Brain_Tumor

# DBSCAN parameters (can be tuned)
eps = 4.0
min_samples = 5

def extract_border_points(X_class):
    dbscan = DBSCAN(eps=eps, min_samples=min_samples)
    dbscan.fit(X_class)

    # core samples mask
    core_mask = np.zeros_like(dbscan.labels_, dtype=bool)
    core_mask[dbscan.core_sample_indices_] = True

    # border points = points that are not core and not noise (-1)
    border_mask = (dbscan.labels_ != -1) & (~core_mask)
    border_points = X_class[border_mask]

    return border_points

border_points_class0 = extract_border_points(X_train_class0)
border_points_class1 = extract_border_points(X_train_class1)

X_border_combined = np.vstack((border_points_class0, border_points_class1))
y_border_combined = np.array(
    [0] * len(border_points_class0) + [1] * len(border_points_class1)
)

print("Border objects extracted:")
print(f"Class 0: {len(border_points_class0)} border points")
print(f"Class 1: {len(border_points_class1)} border points")
print(f"Total after numerosity reduction: {len(X_border_combined)} samples")
```

```python


def evaluate_classifiers_with_time(X_train, y_train, X_test, y_test):
    times = {}

    # KNN
    print("KNN Classifier Performance:")
    for k in [1, 3, 5, 7]:
        knn = KNeighborsClassifier(n_neighbors=k)
        start = time.time()
        knn.fit(X_train, y_train)
        end = time.time()
        train_time = end - start

        y_pred = knn.predict(X_test)
        acc = accuracy_score(y_test, y_pred)
        print(f"  k={k}: Accuracy = {acc:.4f}, Training Time = {train_time:.4f} seconds")
        times[f'knn_{k}'] = train_time

    # Decision Tree
    print("\nDecision Tree Classifier:")
    dt = DecisionTreeClassifier(random_state=42)
    start = time.time()
    dt.fit(X_train, y_train)
    end = time.time()
    dt_time = end - start

    y_pred_dt = dt.predict(X_test)
    acc_dt = accuracy_score(y_test, y_pred_dt)
    print(f"  Accuracy = {acc_dt:.4f}, Training Time = {dt_time:.4f} seconds")
    times['decision_tree'] = dt_time

    # SVM
    print("\nSVM Classifier:")
    svm = SVC(random_state=42)
    start = time.time()
    svm.fit(X_train, y_train)
    end = time.time()
    svm_time = end - start

    y_pred_svm = svm.predict(X_test)
    acc_svm = accuracy_score(y_test, y_pred_svm)
    print(f"  Accuracy = {acc_svm:.4f}, Training Time = {svm_time:.4f} seconds")
    print(f"  Number of Support Vectors = {sum(svm.n_support_)}")
    times['svm'] = svm_time

    return times

print("Evaluating on original data:")
times_original = evaluate_classifiers_with_time(X_train_flat, y_train, X_test_flat, y_test)

print("\nEvaluating on PCA-reduced data:")
times_pca = evaluate_classifiers_with_time(X_train_pca, y_train, X_test_pca, y_test)

print("\nSpeed Up in Training Time After PCA:")
for key in times_original.keys():
    speed_up = times_original[key] / times_pca[key]
    print(f"{key}: Speed up = {speed_up:.2f}x")
```

```python
import matplotlib.pyplot as plt

classifiers = list(times_original.keys())
times_orig = [times_original[c] for c in classifiers]
times_pca = [times_pca[c] for c in classifiers]

x = range(len(classifiers))

plt.figure(figsize=(10,6))
plt.bar([i - 0.15 for i in x], times_orig, width=0.3, label='Original Data')
plt.bar([i + 0.15 for i in x], times_pca, width=0.3, label='PCA Reduced Data')
plt.xticks(x, classifiers)
plt.ylabel('Training Time (seconds)')
plt.title('Training Time Comparison: Original vs PCA Reduced Data')
plt.legend()
plt.grid(True)
plt.show()
```

```python
components_range = range(10, 101, 10)
results_summary = []

best_acc = 0
best_config = None

for n_comp in components_range:
    print(f"\nEvaluating PCA with {n_comp} components...")
    pca = PCA(n_components=n_comp)
    X_train_pca = pca.fit_transform(X_train_flat)
    X_test_pca = pca.transform(X_test_flat)


    results = evaluate_classifiers_with_time(X_train_pca, y_train, X_test_pca, y_test)

    for clf_name, metrics in results.items():
        acc = metrics['accuracy']
        train_time = metrics['train_time']
        print(f"  {clf_name}: Accuracy = {acc:.4f}, Training Time = {train_time:.4f} seconds")
        results_summary.append({
            'classifier': clf_name,
            'n_components': n_comp,
            'accuracy': acc,
            'train_time': train_time
        })

        if acc > best_acc:
            best_acc = acc
            best_config = (clf_name, n_comp, acc, train_time)

print("\nBest classifier configuration:")
print(f"Classifier: {best_config[0]}, PCA components: {best_config[1]}, Accuracy: {best_config[2]:.4f}, Training Time: {best_config[3]:.4f} seconds")

# رسم النتائج
import matplotlib.pyplot as plt

plt.figure(figsize=(12,7))
for clf_name in set(r['classifier'] for r in results_summary):
    x = [r['n_components'] for r in results_summary if r['classifier'] == clf_name]
    y = [r['accuracy'] for r in results_summary if r['classifier'] == clf_name]
    plt.plot(x, y, marker='o', label=clf_name)

plt.xlabel('Number of PCA Components')
plt.ylabel('Accuracy')
plt.title('Classifier Accuracy vs Number of PCA Components')
plt.legend()
plt.grid(True)
plt.show()
```
